```python
# import pandas library
import pandas as pd
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Path to dataset on Google Drive
data_path= "/content/drive/MyDrive/Colab Notebooks/loan_approval_dataset.csv"
```

```python
# Load dataset
loan = pd.read_csv(data_path)
```

```python
# Remove leading and trailing spaces from column names
loan.columns = loan.columns.str.strip()
```

```python
loan[["education", "self_employed", "loan_status"]].value_counts()
```

```python
missing_values = loan.isnull().sum()
print("Missing values in each column:\n", missing_values)
```

```python
# Checking unique values in each column for possible missing value indicators
for col in loan.columns:
    print(f"Unique values in {col}:", loan[col].unique(), "\n")
```

```python
loan['education'] = loan['education'].str.strip()
loan['self_employed'] = loan['self_employed'].str.strip() #remove Whitespace
```

```python
loan.hist(bins=50,figsize=(20,15))
plt.suptitle('Figure 1', fontsize=20)
plt.show()
```

```python
loan.hist(figsize=(10, 14), bins=20)
plt.show()#Histogram (for numerical attributes)
```

```python
## check for outliers
plt.figure(figsize=(12, 6))
sns.boxplot(data=loan.select_dtypes(include=['int64', 'float64']))
plt.xticks(rotation=90)
plt.show()
```

```python
#Scatter plot (loan_amount vs income_annum)
plt.figure(figsize=(8, 6))
sns.scatterplot(x="loan_amount", y="income_annum", data=loan, hue="loan_status")
plt.title("loan_amount vs income_annum")
plt.show()
```

```python
import matplotlib.pyplot as plt
loan_status_counts = loan["loan_status"].value_counts()
```

```python
# Bar Plot
plt.figure(figsize=(6,4))
loan_status_counts.plot(kind='bar', color=['green', 'red'])
plt.xlabel("Loan Status")
plt.ylabel("Count")
plt.title("Loan Approval vs. Rejection")
plt.show()
```
